```python
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# 1. Cargar datos y limpieza
df_cancer_rf = pd.read_csv('breast-cancer.xls')
df_cancer_rf = df_cancer_rf.drop(columns=['id'])

# 2. Separar variables
X_rf = df_cancer_rf.drop(columns=['diagnosis'])
y_rf = df_cancer_rf['diagnosis']

# 3. Dividir datos en entrenamiento (70%) y prueba (30%)
X_train_rf, X_test_rf, y_train_rf, y_test_rf = train_test_split(X_rf, y_rf, test_size=0.3, random_state=42)

# 4. Optimización con GridSearchCV (Buscamos cantidad de árboles y profundidad)
parametros_rf = {
    'n_estimators': [50, 100, 150], # Cuántos árboles tendrá el bosque
    'max_depth': [None, 3, 5, 7]    # Profundidad máxima de cada árbol
}

rf_base = RandomForestClassifier(random_state=42)
grid_rf = GridSearchCV(rf_base, parametros_rf, cv=5)
grid_rf.fit(X_train_rf, y_train_rf)

# 5. Evaluación del mejor modelo
mejor_rf = grid_rf.best_estimator_
y_pred_rf = mejor_rf.predict(X_test_rf)

# 6. Imprimir Métricas
print("--- RESULTADOS RANDOM FOREST ---")
print(f"Mejores hiperparámetros encontrados: {grid_rf.best_params_}")
print(f"Precisión Global (Accuracy): {accuracy_score(y_test_rf, y_pred_rf):.3f}")
print("\nReporte de Clasificación:")
print(classification_report(y_test_rf, y_pred_rf))
```
